# <b>A. <span style='color:#0B2F9F'><code>AGGREGATION</code></b>

Fungsi agregat atau <b>Aggregation</b> adalah fungsi yang melakukan perhitungan pada sekumpulan nilai dan mengembalikan satu nilai hasil perhitungan tersebut. Fungsi agregat SQL yang paling umum digunakan adalah:
<br><br>
<table border="1">
    <tr>
        <th>Fungsi Agregat</th>
        <th>Deskripsi</th>
    </tr>
    <tr>
        <td><b>MIN( )</b></td>
        <td>Mengembalikan <span style='color:#FFB200'>nilai terkecil</span> dalam kolom yang dipilih.</td>
    </tr>
    <tr>
        <td><b>MAX( )</b></td>
        <td>Mengembalikan <span style='color:#FFB200'>nilai terbesar</span> dalam kolom yang dipilih.</td>
    </tr>
    <tr>
        <td><b>COUNT( )</b></td>
        <td>Mengembalikan <span style='color:#FFB200'>jumlah baris</span> dalam suatu set</td>
    </tr>
    <tr>
        <td><b>SUM( )</b></td>
        <td>Mengembalikan <span style='color:#FFB200'>jumlah total</span> dari kolom numerik.</td>
    </tr>
    <tr>
        <td><b>AVG( )</b></td>
        <td>Mengembalikan <span style='color:#FFB200'>nilai rata-rata</span> dari kolom numerik.</td>
    </tr>
</table>


#### <i><b><span style='color:#55679C'>Quest</span> : Hitung banyak data pada tabel telco_services!</b></i>

```python
SELECT COUNT(*) AS jumlah_data FROM telco_churn.telco_services
```

Query tersebut berfungsi untuk menghitung total baris data pada tabel `telco_churn.telco services` menggunakan fungsi agregasi `COUNT(*)`, yang hasilnya diberi nama alias kolom `jumlah_data`. Berdasarkan bagian *Query results*, eksekusi perintah tersebut berhasil diselesaikan dan menunjukkan bahwa tabel tersebut memiliki total data keseluruhan sebanyak **7.043 baris**.

#### <i><b><span style='color:#55679C'>Quest</span> : Hitung dan tampilkan total_revenue pada  tabel telco_services yang didapat perusahaan!</b></i>

```python
SELECT SUM(total_revenue) FROM `telco_churn.telco services`
```

Query tersebut berfungsi untuk **menghitung jumlah total pendapatan (total revenue)** secara keseluruhan dari tabel `telco_churn.telco services`.

Melalui fungsi agregasi `SUM(total_revenue)`, sistem akan menjumlahkan seluruh nilai numerik yang ada pada kolom `total_revenue` untuk menghasilkan satu angka total pendapatan.

#### <i><b><span style='color:#55679C'>Quest</span> : Hitung dan tampilkan nilai minimum, nilai maksimum dan rata-rata kolom tenure_in_months pada tabel telco_services dan beri nama alias yang sesuai!</b></i>

```python
SELECT MIN(tenure_in_months) AS min_tenure,
 MAX(tenure_in_months) AS max_tenure, AVG(tenure_in_months) AS avg_tenure
FROM `telco_churn.telco services`
```

Query tersebut digunakan untuk menghitung statistik lama berlangganan pelanggan (`tenure_in_months`) pada tabel `telco_services` menggunakan fungsi agregasi:

* `MIN` untuk mencari lama berlangganan paling rendah (`min_tenure`)
* `MAX` untuk mencari lama berlangganan paling tinggi (`max_tenure`)
* `AVG` untuk menghitung rata-rata lama berlangganan pelanggan (`avg_tenure`)

Interpretasinya, hasil query memberikan gambaran umum mengenai durasi pelanggan menggunakan layanan perusahaan. Nilai minimum menunjukkan pelanggan dengan masa berlangganan paling singkat, nilai maksimum menunjukkan pelanggan paling lama bertahan, sedangkan rata-rata tenure membantu perusahaan memahami tingkat loyalitas pelanggan secara keseluruhan. Informasi ini dapat digunakan untuk analisis retensi pelanggan dan evaluasi strategi pelayanan perusahaan.


#### <i><b><span style='color:#55679C'>Quest</span> : Buat 3 kolom yang diberi total_offer_A_or_B, total_offer_C_or_D_or_E dan non_offer yang merupakan perhitungan</b></i>
<ul>
<li>total_offer_A_or_B = Jumlah customer yang mendapatkan penawaran Offer A atau Offer B</li>
<li>total_offer_C_or_D_or_E = Jumlah customer yang mendapatkan penawaran Offer C, Offer D atau Offer E</li>
<li>non_offer = Jumlah customer yang tidak mendapatkan penawaran</li>
</ul>

```python
SELECT
COUNT(CASE WHEN offer IN ('Offer A', 'Offer B') THEN offer
END) AS total_offer_A_or_B,
COUNT(CASE WHEN offer IN ('Offer C', 'Offer D', 'Offer E') THEN offer
END) AS total_offer_C_or_D_or_E,
COUNT(CASE WHEN offer IS NULL THEN 2
END) AS non_offer FROM `telco_churn.telco services`
```

Query tersebut digunakan untuk menghitung jumlah pelanggan berdasarkan kategori penawaran (`offer`) menggunakan fungsi `COUNT` dan `CASE`. Hasil perhitungan dibagi menjadi tiga kelompok, yaitu:

* `total_offer_A_or_B` → jumlah pelanggan yang menggunakan `Offer A` atau `Offer B`
* `total_offer_C_or_D_or_E` → jumlah pelanggan yang menggunakan `Offer C`, `Offer D`, atau `Offer E`
* `non_offer` → jumlah pelanggan yang tidak memiliki penawaran (`offer IS NULL`)

Interpretasinya, query ini membantu perusahaan dalam mengetahui distribusi penggunaan program penawaran oleh pelanggan. Dengan data tersebut, perusahaan dapat menganalisis tingkat penggunaan setiap kategori promo, membandingkan efektivitas penawaran, serta mengetahui jumlah pelanggan yang belum mendapatkan atau menggunakan program promo tertentu. Informasi ini berguna untuk evaluasi strategi pemasaran dan pengambilan keputusan bisnis.

# <b>B. <span style='color:#0B2F9F'><code>GROUP BY</code></b>

Fungsi agregat sering digunakan bersama dengan klausa GROUP BY pada pernyataan SELECT. Klausa GROUP BY membagi hasil query menjadi beberapa kelompok nilai, dan fungsi agregat dapat digunakan untuk mengembalikan satu nilai untuk setiap kelompok.

#### <i><b><span style='color:#55679C'>Quest</span> : Hitung dan tampilkan banyaknya customer dari tiap internet_type pada tabel telco_services? Urutkan dari yang jumlahnya terbesar hingga terkecil!</b></i>

```python
SELECT internet_type, COUNT(customer_id) AS jumlah_customer
FROM `telco_churn.telco services`
GROUP BY internet_type ORDER BY jumlah_customer DESC
```

Query tersebut digunakan untuk menghitung jumlah pelanggan (`COUNT(customer_id)`) berdasarkan jenis internet (`internet_type`) pada tabel `telco_services`. Data kemudian dikelompokkan menggunakan `GROUP BY internet_type` dan diurutkan dari jumlah pelanggan terbanyak ke paling sedikit menggunakan `ORDER BY jumlah_customer DESC`.

Interpretasinya, hasil query menunjukkan distribusi pelanggan berdasarkan jenis layanan internet yang digunakan, seperti Fiber Optic, DSL, atau Cable. Dengan pengurutan secara descending, perusahaan dapat mengetahui jenis internet yang paling banyak digunakan pelanggan. Informasi ini bermanfaat untuk menganalisis popularitas layanan internet, menentukan prioritas pengembangan layanan, serta mendukung strategi pemasaran dan peningkatan kualitas jaringan.


#### <i><b><span style='color:#55679C'>Quest</span> : Hitung dan tampilkan banyaknya customer tiap kategori offer dan internet_type pada tabel telco_services? Urutkan berdasarkan offer dan internet_type secara ascending!</b></i>

```python
SELECT offer, internet_type, COUNT(customer_id) AS jumlah_customer FROM `telco_churn.telco services`
GROUP BY offer, internet_type ORDER BY offer ASC, internet_type ASC
```

Query tersebut digunakan untuk menghitung jumlah pelanggan (`COUNT(customer_id)`) berdasarkan kombinasi jenis penawaran (`offer`) dan jenis internet (`internet_type`) pada tabel `telco_services`. Data dikelompokkan menggunakan `GROUP BY offer, internet_type` dan diurutkan secara ascending berdasarkan nama `offer` dan `internet_type`.

Interpretasinya, hasil query menunjukkan jumlah pelanggan pada setiap kombinasi penawaran dan layanan internet yang digunakan. Misalnya, perusahaan dapat mengetahui berapa banyak pelanggan pengguna `Offer A` yang memakai `Fiber Optic`, `DSL`, atau jenis internet lainnya. Informasi ini membantu perusahaan dalam menganalisis hubungan antara program promo dengan pilihan layanan internet pelanggan, sehingga dapat digunakan untuk evaluasi strategi pemasaran, segmentasi pelanggan, dan pengembangan layanan yang lebih tepat sasaran.


#### <i><b><span style='color:#55679C'>Quest</span> : Tampilkan kolom phone_internet_type dan total_customer tiap tipe tersebut pada tabel telco_services lalu urutkan dari total_customer terbesar hingga terkecil, dengan ketentuan</b></i><br>
<ul>
<li>Jika customer menggunakan layanan Telepon & Internet, maka phone_internet_type = Phone & Internet</li>
<li>Jika customer menggunakan layanan Telepon namun tidak menggunakan layanan Internet, maka phone_internet_type = Phone Only</li>
<li>Jika customer tidak menggunakan layanan Telepon namun menggunakan layanan Internet, maka phone_internet_type = Internet Only</li>
<li>Jika customer tidak menggunakan kedua layanan tersebut, maka phone_internet_type = No Phone & Internet</li>
</ul>

```python
SELECT
CASE
WHEN phone_service = TRUE AND internet_service = TRUE THEN  'PHONE & INTERNET'
WHEN phone_service = TRUE AND internet_service = FALSE  THEN  'PHONE ONLY'
WHEN internet_service = TRUE AND  phone_service = FALSE THEN  'INTERNET  ONLY'
ELSE 'NO PHONE & INTERNET'
END AS phone_internet_type,
COUNT (customer_id) AS jumlah_customer
FROM `telco_churn.telco services`
GROUP BY phone_internet_type
ORDER BY jumlah_customer DESC
```

Query tersebut digunakan untuk mengelompokkan pelanggan berdasarkan kombinasi penggunaan layanan telepon (`phone_service`) dan layanan internet (`internet_service`) menggunakan fungsi `CASE`. Setelah itu, query menghitung jumlah pelanggan (`COUNT(customer_id)`) pada setiap kategori, lalu mengurutkannya dari jumlah terbanyak ke paling sedikit (`ORDER BY jumlah_customer DESC`).

Kategori yang dihasilkan meliputi:

* `PHONE & INTERNET` → pelanggan menggunakan layanan telepon dan internet
* `PHONE ONLY` → pelanggan hanya menggunakan layanan telepon
* `INTERNET ONLY` → pelanggan hanya menggunakan layanan internet
* `NO PHONE & INTERNET` → pelanggan tidak menggunakan kedua layanan

Interpretasinya, hasil query menunjukkan distribusi jumlah pelanggan berdasarkan kombinasi layanan yang digunakan. Dengan data tersebut, perusahaan dapat mengetahui layanan mana yang paling banyak digunakan pelanggan. Informasi ini bermanfaat untuk analisis kebutuhan pelanggan, evaluasi layanan telekomunikasi, serta penyusunan strategi pemasaran dan pengembangan produk yang lebih sesuai dengan pola penggunaan pelanggan.

# <b>C. <span style='color:#0B2F9F'><code>HAVING</code></b>

HAVING dalam SQL digunakan untuk memfilter hasil dari kelompok data yang telah dikelompokkan dengan pernyataan GROUP BY. Mirip dengan klausa WHERE, tetapi HAVING bekerja setelah pengelompokan data dilakukan atau ada agregasi. Dengan kata lain, HAVING biasa digunakan untuk menerapkan kondisi pada hasil agregat.

#### <i><b><span style='color:#55679C'>Quest</span> : Pada tabel telco_services, hitung rata-rata tenure_in_month tiap banyaknya number_of_referrals. Tampilkan data yang rata-rata tenure_in_month kurang dari 40 bulan!</b></i>

```python
SELECT number_of_referrals, AVG(tenure_in_months) AS avg_tenure_in_month
FROM `telco_churn.telco services`
GROUP BY number_of_referrals HAVING avg_tenure_in_month < 40
```

Query tersebut digunakan untuk menghitung rata-rata lama berlangganan pelanggan (`AVG(tenure_in_months)`) berdasarkan jumlah referral (`number_of_referrals`) pada tabel `telco_services`. Data kemudian dikelompokkan menggunakan `GROUP BY number_of_referrals` dan hanya menampilkan kelompok dengan rata-rata lama berlangganan kurang dari 40 bulan (`HAVING avg_tenure_in_month < 40`).

Interpretasinya, hasil query menunjukkan hubungan antara jumlah referral pelanggan dengan rata-rata lama mereka berlangganan. Kelompok pelanggan yang memiliki rata-rata tenure di bawah 40 bulan akan ditampilkan, sehingga perusahaan dapat menganalisis apakah pelanggan dengan jumlah referral tertentu cenderung memiliki masa berlangganan yang lebih singkat. Informasi ini dapat digunakan untuk evaluasi efektivitas program referral dan strategi peningkatan loyalitas pelanggan.


#### <i><b><span style='color:#55679C'>Quest</span> : Hitung dan tampilkan banyaknya customer tiap kategori offer dan internet_type pada tabel telco_services, namun hanya untuk customer yang menerima offer dan total banyak customer lebih dari sama dengan 100 pada kategori yang telah disebutkan!</b></i>

```python
SELECT offer, internet_type, COUNT (customer_id) AS jumlah_customer FROM `telco_churn.telco services`
GROUP BY offer, internet_type
HAVING offer IS NOT NULL AND jumlah_customer >=100
```

Query tersebut digunakan untuk menghitung jumlah pelanggan (`COUNT(customer_id)`) berdasarkan kombinasi jenis penawaran (`offer`) dan jenis internet (`internet_type`) pada tabel `telco_services`. Data dikelompokkan menggunakan `GROUP BY offer, internet_type`, kemudian hanya menampilkan data dengan kondisi `offer` tidak kosong (`offer IS NOT NULL`) dan jumlah pelanggan minimal 100 (`jumlah_customer >= 100`) melalui klausa `HAVING`.

Interpretasinya, hasil query menunjukkan kombinasi penawaran dan jenis internet yang memiliki jumlah pelanggan cukup besar, yaitu minimal 100 pelanggan. Data ini membantu perusahaan dalam mengidentifikasi program penawaran dan layanan internet yang paling banyak diminati pelanggan. Informasi tersebut dapat dimanfaatkan untuk evaluasi efektivitas promosi, pengembangan layanan unggulan, serta penyusunan strategi pemasaran yang lebih tepat sasaran.
